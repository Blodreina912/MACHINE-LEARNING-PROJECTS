# Task
Analyze the provided student study hours and scores data. Create a CSV file named "student_scores.csv" with the data. Load the data from the CSV file. Visualize the data using a scatter plot. Split the data into training and testing sets. Train a Simple Linear Regression model. Predict test set results. Evaluate model performance using Mean Absolute Error (MAE) and R² Score. Plot the regression line along with data points. Predict the score for a student studying 9.25 hours.

## Create the dataset

### Subtask:
Create a CSV file named `student_scores.csv` with the provided data.


**Reasoning**:
Create a pandas DataFrame and save it to a CSV file as instructed.



```python
import pandas as pd

data = {'Hours': [2.5, 5.1, 3.2, 8.5, 3.5, 1.5, 9.2, 5.5, 8.3, 2.7, 7.7, 5.9, 4.5, 3.3, 1.1, 8.9, 2.5, 1.9, 6.1, 7.4, 2.7, 4.8, 3.8, 6.9, 7.8],
        'Scores': [21, 47, 27, 75, 30, 20, 88, 60, 81, 25, 85, 62, 41, 42, 17, 95, 30, 24, 67, 69, 30, 54, 35, 76, 86]}

df = pd.DataFrame(data)
df.to_csv('student_scores.csv', index=False)
```

## Load the dataset

### Subtask:
Load the data from `student_scores.csv` into a pandas DataFrame.


**Reasoning**:
Import the pandas library and load the data from the CSV file into a DataFrame.



```python
df = pd.read_csv('student_scores.csv')
display(df.head())
```

## Visualize the data

### Subtask:
Create a scatter plot to visualize the relationship between study hours and scores.


**Reasoning**:
Create a scatter plot to visualize the relationship between study hours and scores, adding labels and a title.



```python
import matplotlib.pyplot as plt

plt.scatter(df['Hours'], df['Scores'])
plt.title('Study Hours vs. Scores')
plt.xlabel('Hours Studied')
plt.ylabel('Score')
plt.show()
```

## Split the data

### Subtask:
Split the data into training and testing sets for the linear regression model.


**Reasoning**:
Split the data into training and testing sets using train_test_split.



```python
from sklearn.model_selection import train_test_split

X = df['Hours'].values.reshape(-1, 1)
y = df['Scores']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

## Train the model

### Subtask:
Train a Simple Linear Regression model on the training data.


**Reasoning**:
Train a Simple Linear Regression model on the training data.



```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
```

## Predict test results

### Subtask:
Use the trained model to predict scores on the test data.


**Reasoning**:
Use the trained model to predict scores on the test data.



```python
y_pred = model.predict(X_test)
```

## Evaluate the model

### Subtask:
Calculate and display the Mean Absolute Error (MAE) and R² Score to evaluate the model's performance.


**Reasoning**:
Calculate and display the Mean Absolute Error and R² Score using the actual and predicted test scores.



```python
from sklearn.metrics import mean_absolute_error, r2_score

mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f'Mean Absolute Error (MAE): {mae:.2f}')
print(f'R² Score: {r2:.2f}')
```

## Visualize the regression line

### Subtask:
Plot the regression line on the scatter plot along with the data points.


**Reasoning**:
Create a scatter plot of the training data and plot the regression line on the same plot.



```python
plt.scatter(X_train, y_train, color='blue', label='Training Data')
plt.plot(X_train, model.predict(X_train), color='red', label='Regression Line')
plt.title('Study Hours vs. Scores (Training set) with Regression Line')
plt.xlabel('Hours Studied')
plt.ylabel('Score')
plt.legend()
plt.show()
```

## Predict for a new value

### Subtask:
Predict the score for a student studying 9.25 hours using the trained model.


**Reasoning**:
Predict the score for a student studying 9.25 hours using the trained model.



```python
import numpy as np

hours_to_predict = np.array([9.25]).reshape(-1, 1)
predicted_score = model.predict(hours_to_predict)

print(f'Predicted score for a student studying 9.25 hours: {predicted_score[0]:.2f}')
```
