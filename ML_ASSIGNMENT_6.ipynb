```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris, load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
from sklearn.metrics import accuracy_score, mean_squared_error

# --- Classification Example (Iris Dataset) ---

print("--- KNN Classification (Iris Dataset) ---")

# Load the Iris dataset
iris = load_iris()
X_iris, y_iris = iris.data, iris.target

# Split the dataset into training and testing sets
X_train_iris, X_test_iris, y_train_iris, y_test_iris = train_test_split(
    X_iris, y_iris, test_size=0.3, random_state=42
)

# Initialize and train the KNN classifier
knn_classifier = KNeighborsClassifier(n_neighbors=3)
knn_classifier.fit(X_train_iris, y_train_iris)

# Make predictions and evaluate the classifier
y_pred_iris = knn_classifier.predict(X_test_iris)
accuracy = accuracy_score(y_test_iris, y_pred_iris)
print(f"Accuracy of KNN Classifier: {accuracy:.2f}")

# --- Regression Example (Diabetes Dataset) ---

print("\n--- KNN Regression (Diabetes Dataset) ---")

# Load the Diabetes dataset
diabetes = load_diabetes()
X_diabetes, y_diabetes = diabetes.data, diabetes.target

# Split the dataset into training and testing sets
X_train_diabetes, X_test_diabetes, y_train_diabetes, y_test_diabetes = train_test_split(
    X_diabetes, y_diabetes, test_size=0.3, random_state=42
)

# Initialize and train the KNN regressor
knn_regressor = KNeighborsRegressor(n_neighbors=5)
knn_regressor.fit(X_train_diabetes, y_train_diabetes)

# Make predictions and evaluate the regressor
y_pred_diabetes = knn_regressor.predict(X_test_diabetes)
mse = mean_squared_error(y_test_diabetes, y_pred_diabetes)
print(f"Mean Squared Error of KNN Regressor: {mse:.2f}")

# Optional: Visualize the results for the classification example (first two features)
# This is a simplified visualization for demonstration purposes
plt.figure(figsize=(10, 6))
plt.scatter(X_test_iris[:, 0], X_test_iris[:, 1], c=y_test_iris, cmap='viridis', marker='o', label='True Labels')
plt.scatter(X_test_iris[:, 0], X_test_iris[:, 1], c=y_pred_iris, cmap='plasma', marker='x', s=100, label='Predictions')
plt.title('KNN Classification on Iris Dataset (First Two Features)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```
